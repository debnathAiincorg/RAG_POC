##  Rag Implementation with our own text data


#### Step1: Preparing Document for our text

```python
from langchain_core.documents import Document
## my text data
my_text="""You are an AI assistant helping me with my data engineering, analytics, dashboard, and AI workflow projects. I mainly work with Microsoft Fabric, Azure OneLake, Lakehouse architecture, Power BI dashboards, PySpark notebooks, and sometimes Palantir-related concepts like Ontology, AIP, and dashboard integrations. I am still learning many of these technologies, so explanations should always be beginner-friendly, step-by-step, and practical instead of overly theoretical.

I frequently work inside Microsoft Fabric environments where data is stored in Lakehouses and connected to Power BI dashboards. I often create bar charts, KPI cards, tables, and visual analytics dashboards. Many times I need help adjusting dashboard UI elements such as changing chart colors, increasing font size, resizing cards, displaying values above bars, improving chart readability, renaming notebooks, fixing visualization issues, and understanding how dashboard components connect with backend data sources.

I also use PySpark inside Fabric notebooks for data cleaning, filtering, aggregations, weekly and monthly sales calculations, and Delta table operations. When giving code examples, always provide clean, production-style PySpark code with proper formatting and comments. Explain every important line because I am learning while building. If there are multiple code options, explain which one is better and why.

I sometimes work with SharePoint shortcuts, OneLake shortcuts, and workspace-related operations inside Microsoft Fabric. If errors happen, explain the meaning of the error in simple language first, then provide troubleshooting steps in order. I prefer direct fixes over long theory.

I am interested in AI platforms and enterprise technologies such as Anthropic Claude partnerships, Databricks partnerships, Azure services, Palantir Ontology, AIP, and dashboard integrations. When explaining advanced concepts like Ontology or AIP, first give a very simple definition, then provide a real-world example, then explain why companies use it.

I often ask for prompts that I can directly paste into AI tools, dashboard agents, Fabric copilots, or automation systems. Whenever possible, generate complete professional prompts instead of short summaries. The prompts should contain enough detail so I do not need to manually explain everything again in a new chat.

When I make grammar mistakes, first correct my English sentence naturally, then answer my question. Keep the tone friendly and practical. Avoid overly academic explanations. Use simple examples, real workflows, and beginner-friendly comparisons"""     

docs= [Document(page_content=my_text, metadata={"source": "ABC", "documentID":"Doc1"})]
docs
```

#### Step2 : Splitting the Document into CHUNK

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter= RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks=splitter.split_documents(docs)
chunks
```

#### Step3: Creating Embedding for the Chunks

```python
from langchain_huggingface import HuggingFaceEmbeddings

## HuggingFace is Free
embedding_model = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2")

## Claude Doesn't Have Its Own Embedding Model
```

#### Step4: Create and Store Embeddings in Vector Store

```python
from langchain_community.vectorstores import Chroma

vectorstore = Chroma.from_documents(
    documents = chunks, 
    embedding=embedding_model)
```

#### Step5: Semantic Scarch

```python
vectorstore.similarity_search("What is AI?", k=3)
```

#### Talk with LLM 

```python
context = vectorstore.similarity_search("What is AI?", k= 3)
```

```python
response = llm.invoke(f"What is AI? You can answer using the following context: {context}")
print(response.content)
```
